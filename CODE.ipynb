```python
import pandas as pd

# Replace with your uploaded file name
file_name = "/content/DecodeX_VoltRide_Dataset (1).xlsx"

df = pd.read_excel(file_name)

# Preview columns
print(df.columns)
df.head()

```

```python
# Remove missing pickup zones if any
df = df.dropna(subset=["Pickup_Zone"])

# Convert Hour column to numeric if needed
df["Hour"] = pd.to_numeric(df["Hour"], errors="coerce")
```

```python
import matplotlib.pyplot as plt

demand_zone = df["Pickup_Zone"].value_counts()

plt.figure(figsize=(10,5))
demand_zone.plot(kind="bar")
plt.title("Demand by Pickup Zone")
plt.xlabel("Pickup Zone")
plt.ylabel("Number of Rides")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
hourly_demand = df["Hour"].value_counts().sort_index()

plt.figure(figsize=(10,5))
hourly_demand.plot(kind="line", marker="o")
plt.title("Hourly Ride Demand")
plt.xlabel("Hour")
plt.ylabel("Number of Rides")
plt.tight_layout()
plt.show()
```

```python
ride_status = df["Ride_Status"].value_counts()

plt.figure(figsize=(6,6))
ride_status.plot(kind="pie", autopct="%1.1f%%")
plt.title("Ride Completion vs Cancellation")
plt.ylabel("")
plt.show()
```

```python
city_demand = df["City"].value_counts()

plt.figure(figsize=(8,5))
city_demand.plot(kind="bar")
plt.title("Demand by City")
plt.xlabel("City")
plt.ylabel("Number of Rides")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
surge_dist = df["Surge_Multiplier"].value_counts().sort_index()

plt.figure(figsize=(8,5))
surge_dist.plot(kind="bar")
plt.title("Surge Multiplier Distribution")
plt.xlabel("Surge Multiplier")
plt.ylabel("Number of Rides")
plt.tight_layout()
plt.show()
```
